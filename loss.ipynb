```python
import numpy as np
```

```python
y_predicted = np.array([1,1,0,0,1])
y_true = np.array([0.3,0.7,1,0,0.5])
```

```python
def mae(y_true, y_predicted):
    return np.mean(np.abs(y_true - y_predicted))
```

```python
mae(y_true, y_predicted)
```

```python
def mse(y_true, y_predicted):
    return np.mean((y_true - y_predicted)**2)
```

```python
mse(y_true, y_predicted)
```

```python
def log_loss(y_true, y_predicted):
    if np.any(y_predicted == 0) or np.any(y_predicted == 1):
        return -np.mean(y_true * np.log(np.clip(y_predicted, 1e-15, 1 - 1e-15)) + (1-y_true) * np.log(np.clip(1-y_predicted, 1e-15, 1 - 1e-15)))
    else:
        return -np.mean(y_true * np.log(y_predicted) + (1-y_true) * np.log(1-y_predicted))
```

```python
log_loss(y_true, y_predicted)
```

```python
def mse_without_np(y_true, y_predicted):
    n = len(y_true)
    mse_sum = 0
    for i in range(n):
        mse_sum += (y_true[i] - y_predicted[i]) ** 2
    return mse_sum / n
```

```python
mse_without_np(y_true, y_predicted)
```
